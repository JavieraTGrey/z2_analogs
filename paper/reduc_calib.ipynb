```python
import numpy as np
from astropy.io import fits
from scipy.interpolate import griddata

hdul = hdu5

flux = hdul['DET01-SCIIMG'].data
ivar = hdul['DET01-IVARRAW'].data
wave = hdul['DET01-WAVEIMG'].data
slitmask = hdu5[10].data['spat_id']

# Select one slit
slit_id = 626
gpm = slitmask == slit_id

# Extract coordinates
y, x = np.indices(flux.shape)

# Flatten valid pixels
xv = x[gpm]
yv = y[gpm]
wv = wave[gpm]
fv = flux[gpm]

# Define linear wavelength grid
wave_grid = np.linspace(np.nanmin(wv), np.nanmax(wv), 2000)
spat_grid = np.linspace(yv.min(), yv.max(), 200)

W, Y = np.meshgrid(wave_grid, spat_grid)

# Interpolate
rectified = griddata(
    (wv, yv),
    fv,
    (W, Y),
    method='linear'
)
```
