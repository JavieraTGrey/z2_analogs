```python
def save_fluxes_au():
    DIR = '/Users/javieratoro/Desktop/thesis/proyecto 2024-2/'
    lines = pd.read_csv(DIR + 'CSV_files/emission_line1.csv')

    names = ['J0023', 'J0136', 'J0020', 'J0203', 'J0243', 'J0333',
             'J0404','J2204', 'J2258', 'J2336', 'J0328']

    columns_ = ['ID', 'mass', 'z_red', 'z_blue']
    for line in lines['name']:
        columns_.append(line + '_flux')
        columns_.append(line + '_fluxerr')

    df = pd.DataFrame(columns=columns_)

    bright_lines = ['O2_3725', 'O2_3727', 'H_alpha', 'H_beta',
                    'H_gamma', 'O3_5008', 'O3_4959', 'N2_6550',
                    'N2_6585', 'S2_6716', 'S2_6730']

    auroral_lines = ['N2_5756', 'O1_6363',
                     'O3_4363', 'S3_6312',
                     'O2_7322', 'O2_7333',
                     'O2_7322_7333', 'S2_4068', 'S2_4076']

    for name in names:
        fluxes = pd.read_csv(DIR + f'lines/{name}_line_fluxes_parts.csv')
        data = pd.read_csv(DIR + f'lines/{name}/{name}_model_parts.csv')
        auroral = pd.read_csv(DIR +f'lines/{name}/{name}_auroral_model_parts.csv')
        all_rows = {'ID':data['ID'][0], 'mass':data['mass'][0],
                    'z_red':data['z_red'][0], "z_blue":data['z_blue'][0]}
        for line in lines['name']:
            if line in auroral_lines:
                print(line)
                flux = auroral[auroral['name']==line]['flux'].values[0]
                flux_err = auroral[auroral['name']==line]['fluxerr'].values[0]
                print(flux, flux_err)
                all_rows[line + '_flux'] = flux
                all_rows[line + '_fluxerr'] = flux_err
            else:
                narrow = fluxes[line + '_narrow'].values
                flux = narrow
                if line in bright_lines:
                    broad = fluxes[line + '_broad'].values
                    flux += broad
                all_rows[line + '_flux'] = np.median(flux)
                all_rows[line + '_fluxerr'] = np.std(flux)

        df.loc[-1] = all_rows
        df.index = df.index + 1
    df.to_csv(DIR + 'lines/magE2024_master_au_parts.csv')
    return df
```

```python
def save_fluxes_or():
    DIR = '/Users/javieratoro/Desktop/thesis/proyecto 2024-2/'
    lines = pd.read_csv(DIR + 'CSV_files/emission_line1.csv')

    names = ['J0023', 'J0136', 'J0020', 'J0203', 'J0243', 'J0333',
             'J0404', 'J2204', 'J2258', 'J2336', 'J0328']

    columns_ = ['ID', 'mass', 'z_red', 'z_blue']
    for line in lines['name']:
        columns_.append(line + '_flux')
        columns_.append(line + '_fluxerr')

    df = pd.DataFrame(columns=columns_)

    bright_lines = ['O2_3725', 'O2_3727', 'H_alpha', 'H_beta',
                    'H_gamma', 'O3_5008', 'O3_4959', 'N2_6550',
                    'N2_6585', 'S2_6716', 'S2_6730']


    for name in names:
        all_rows = {'ID':name}
        fluxes = pd.read_csv(DIR + f'lines/{name}_line_fluxes_parts.csv')
        data = pd.read_csv(DIR + f'lines/{name}/{name}_model_parts.csv')
        all_rows = {'ID':data['ID'][0], 'mass':data['mass'][0],
                    'z_red':data['z_red'][0], "z_blue":data['z_blue'][0]}
        for line in lines['name']:
            narrow = fluxes[line + '_narrow'].values
            flux = narrow
            if line in bright_lines:
                broad = fluxes[line + '_broad'].values
                flux += broad

            all_rows[line + '_flux'] = np.median(flux)
            all_rows[line + '_fluxerr'] = np.std(flux)

        df.loc[-1] = all_rows
        df.index = df.index + 1
    df.to_csv(DIR + 'lines/magE2024_master_or_parts.csv')
    return df
```
